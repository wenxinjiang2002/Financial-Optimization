```julia
## Problem 4
#### (a)(ii)
GRB_ENV = Gurobi.Env() 
model = Model(() -> Gurobi.Optimizer(GRB_ENV))
set_optimizer_attribute(model, "OutputFlag", 0)
W = 30  
wi = [3, 4, 7, 9, 13, 16] 
num_customers = length(wi)

patterns = []

for i in 1:num_customers
    pattern = zeros(Int, num_customers) 
    pattern[i] = div(W, wi[i]) 
    push!(patterns, pattern)
end

println("Starting Patterns:")
for i in 1:num_customers
    println("Pattern $i: ", patterns[i])
end
```

```julia
## Problem 4
#### (b)
function solve_restricted_problem(w, b, W, patterns)
    model = Model(Gurobi.Optimizer)
    set_optimizer_attribute(model, "OutputFlag", 0) 

    num_patterns = length(patterns)
    num_customers = length(w)
    @variable(model, x[1:num_patterns] >= 0)

    @objective(model, Min, sum(x))

    customer_constraints = []
    for i in 1:num_customers
        con = @constraint(model, sum(patterns[j][i] * x[j] for j in 1:num_patterns) >= b[i])
        push!(customer_constraints, con)
    end
    optimize!(model)
    primal_solution = value.(x)
    dual_solution = [dual(con) for con in customer_constraints]

    return primal_solution, dual_solution
end
w = [3, 4, 7, 9, 13, 16] 
b = [120, 280, 130, 110, 80, 150] 
W = 30  
patterns = [
    [10, 0, 0, 0, 0, 0],  # Pattern 1
    [0, 7, 0, 0, 0, 0],   # Pattern 2
    [0, 0, 4, 0, 0, 0],   # Pattern 3
    [0, 0, 0, 3, 0, 0],   # Pattern 4
    [0, 0, 0, 0, 2, 0],   # Pattern 5
    [0, 0, 0, 0, 0, 1]    # Pattern 6
]

primal_solution, dual_solution = solve_restricted_problem(w, b, W, patterns)

println("Primal Solution (number of rolls for each pattern): ", primal_solution)
println("Total number of rolls used: ", sum(primal_solution))
println("Dual Solution (shadow prices for demand constraints): ", dual_solution)
```

```julia
## Problem 4
#### (c)
function solve_subproblem(w, W, p)
    n = length(w)

     model=Model(()->Gurobi.Optimizer(GRB_ENV))
    set_optimizer_attribute(model, "OutputFlag", 0)  
    @variable(model, x[1:n] >= 0, Int)

    @objective(model, Max, sum(p[i] * x[i] for i in 1:n))
    @constraint(model, sum(w[i] * x[i] for i in 1:n) <= W)

    optimize!(model)

    new_pattern = value.(x)
    optimal_objective = objective_value(model)

    return optimal_objective, new_pattern
end
w = [3, 4, 7, 9, 13, 16]  
W = 30 

p=[0.1,0.14285714,0.25,0.33333333,0.5,1.0]
optimal_objective, new_pattern = solve_subproblem(w, W, p)
reduced_cost=1 - optimal_objective
println("Optimal Objective : ", optimal_objective)
println("Reduced cost of the new pattern: $reduced_cost")
println("New Pattern generated: ", new_pattern)
```

```julia
## Problem 4
#### (d)(i)(ii)
function solve_with_column_generation(w, b, W)
    n = length(w)
    patterns = []
    for i in 1:n
        k = floor(Int, W / w[i])
        a = zeros(Int, n)
        a[i] = k
        push!(patterns, a)
    end

    while true
        primal_sol, duals = solve_restricted_problem(w, b, W, patterns)
        obj_value, new_pattern = solve_subproblem(w, W, duals)
        reduced_cost = 1 - obj_value
        if reduced_cost >= -1e-6
            break
        else
            push!(patterns, new_pattern)
        end
    end
    primal_sol, duals = solve_restricted_problem(w, b, W, patterns)
    return patterns, primal_sol
end
using DataFrames
patterns, primal_sol = solve_with_column_generation(w, b, W)

used_patterns = []
used_pattern_counts = []
for i in 1:length(primal_sol)
    if primal_sol[i] > 1e-5 #
        push!(used_patterns, patterns[i])
        push!(used_pattern_counts, primal_sol[i])
    end
end

pattern_rows = []
for idx in 1:length(used_patterns)
    pattern = used_patterns[idx]
    count = used_pattern_counts[idx]
    pattern_description = ""
    for i in 1:length(w)
        if pattern[i] > 0
            if length(pattern_description) > 0
                pattern_description *= ", "
            end
            pattern_description *= "$(pattern[i]) x width $(w[i])"
        end
    end
    push!(pattern_rows, (idx, count, pattern_description))
end

df = DataFrame("Index" => [row[1] for row in pattern_rows],
               "Times Used" => [row[2] for row in pattern_rows],
               "Pattern Details" => [row[3] for row in pattern_rows])

println("\nOptimal Solution:")
println(df)
integral_pattern_counts = ceil.(used_pattern_counts)
df_integral = DataFrame("Index" => [i for i in 1:length(used_patterns)],
                        "Times Used (Integral)" => integral_pattern_counts,
                        "Pattern Details" => [pattern_rows[i][3] for i in 1:length(pattern_rows)])

println("\nIntegral Solution After Rounding Up:")
println(df_integral)
```
